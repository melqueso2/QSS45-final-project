```python
from pathlib import Path
import pandas as pd

PROJECT_DIR = Path.cwd().parent
DATA_DIR = PROJECT_DIR / "data"
OUTPUT_DIR = PROJECT_DIR / "output"

DATA_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)

print("Project folder:", PROJECT_DIR)
print("Data folder:", DATA_DIR)
print("Output folder:", OUTPUT_DIR)
```

```python
def download_owid(url, filename):
    df = pd.read_csv(
        url,
        storage_options={"User-Agent": "Our World In Data data fetch/1.0"}
    )

    path = DATA_DIR / filename
    df.to_csv(path, index=False)

    print(f"Saved: {filename}")
    print(f"Rows: {len(df):,}")
    print(f"Columns: {df.columns.tolist()}")

    return df
```

```python
WATER_URL = "https://ourworldindata.org/grapher/population-using-at-least-basic-drinking-water.csv?v=1&csvType=full&useColumnShortNames=false"

RENTS_URL = "https://ourworldindata.org/grapher/natural-resource-rents.csv?v=1&csvType=full&useColumnShortNames=false"

GDP_URL = "https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=false"

ELECTRICITY_URL = "https://ourworldindata.org/grapher/share-of-the-population-with-access-to-electricity.csv?v=1&csvType=full&useColumnShortNames=false"
```

```python
water = download_owid(WATER_URL, "water.csv")
rents = download_owid(RENTS_URL, "resource_rents.csv")
gdp = download_owid(GDP_URL, "gdp.csv")
electricity = download_owid(ELECTRICITY_URL, "electricity.csv")
```
